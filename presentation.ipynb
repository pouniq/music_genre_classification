
# پروژه کلاس بندی ژانر موسیقی 
برای رسیدن به این هدف نیاز داریم که داده هایی زیادی در ژانر های متفاوت داشته باشیم دیتاست 

gtzan 

در ده کلاس (ژانر) و هر کدام از کلاس ها ۱۰۰ نمونه موسیقی به مدت ۳۰ ثانیه تهیه کرده است

## تقسیم هر موسیقی به بخش های کوچکتر

هر ترک این دیتاست در قالب موسیقی ۳۰ ثانیه وجود دارد و باید این ۳۰ ثانیه را به بخش های کوچکتری تقسیم کنیم تا بتوانیم الگو را در هر قسمت پیدا کنیم که به این فرآیند

Framing 

گفته می شود

برای اینکه بدانیم هر موسیقی را به چند قسمت چند ثانیه ای تقسیم کنیم نیاز به تکرار و آزمایش کردن داریم

زمانی هم که هر موسیقی را به قسمت های کوچکتر تقسیم می کنیم باید کاری کنیم که از 

data leakage 

جلوگیری شود 

برای این موضوع نیز هر قسمت موسیقی که تقسیم می شود همه قسمت های آن در بخش آموزش باید باشد یا حتما همشون در بخش آزمایش


به عنوان مثال اگر یک صدا را به بخش های ۵ ثانیه ای تقسیم کرده باشم و ۳ تا از بخش ها در قسمت آموزش قرار گرفته باشد و ۳ بخش دیگر آن در قسمت آزمایش باشند به این موضوع 


data leakage 

گفته می شود 

چه مشکلی ایجاد می کند؟

 قسمت‌های یک ترک مستقل از هم نیستند: همان خواننده، همان ساز، همان میکس و ضبط، همان گام و تمپو.
- مدل می‌تواند به‌جای یادگرفتن الگوی ژانر،   ترک را حفظ کند.
- در آزمون، مدل قسمتی از ترکی را می‌بیند که قسمت‌های دیگرش را در آموزش دیده است. پس درست جواب می‌دهد، اما نه چون ژانر را فهمیده، بلکه چون ترک را شناخته است.


برای رفع این مشکل باید هر قسمت از یک ترک حتما در آموزش یا فقط حتما در آزمایش باشد
